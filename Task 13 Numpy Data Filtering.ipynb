# NumPy Data Filtering (Task 13)

**Objective:** Use NumPy conditions to identify values that are above, below, or equal to a specified threshold. Practice Boolean indexing and numerical filtering.

```python
import numpy as np

# Create a NumPy array of student marks
student_marks = np.array([45, 78, 92, 33, 85, 60, 55, 98, 41, 70])
print(f"Original Array (Student Marks): {student_marks}")
```

### 1. Filtering Values Above a Threshold
Let's find all the marks that are greater than 75.

```python
# Create a Boolean mask for marks > 75
high_marks_mask = student_marks > 75

# Use Boolean indexing to filter the array
high_marks = student_marks[high_marks_mask]

print(f"Marks > 75: {high_marks}")
```

### 2. Filtering Values Below a Threshold
Let's find all the marks that are less than 50 (failing grades).

```python
# Shorthand for Boolean indexing (condition inside brackets)
failing_marks = student_marks[student_marks < 50]

print(f"Marks < 50: {failing_marks}")
```

### 3. Filtering Values Equal to a Threshold
Let's check if any student got exactly 60.

```python
exact_60 = student_marks[student_marks == 60]

print(f"Marks equal to 60: {exact_60}")
```

### 4. Combining Conditions (AND / OR)
We can combine conditions using `&` (AND) and `|` (OR). Note the use of parentheses around each condition.
Let's find marks that are between 60 and 80 (inclusive).

```python
average_marks = student_marks[(student_marks >= 60) & (student_marks <= 80)]
print(f"Marks between 60 and 80: {average_marks}")
```

### Interview Questions & Answers

**1. What is Boolean indexing?**
Boolean indexing is a technique in NumPy (and Pandas) where you use an array of Boolean values (True/False) to select elements from another array. The resulting array contains only the elements corresponding to the `True` values in the Boolean mask.

**2. How do you filter a NumPy array?**
You filter a NumPy array by creating a condition that evaluates to a Boolean mask (e.g., `array > 10`), and then passing that mask inside square brackets to index the original array (e.g., `filtered_array = array[array > 10]`).

**3. What is a Boolean array?**
A Boolean array is a NumPy array where every element is either `True` or `False`. They are often the result of applying logical operators (like `>`, `<`, `==`) to a numerical array.
